# Italian Restaurant Sales Analytics

## 05 - Products Data Cleaning and Preparation

### Objective

This notebook cleans and validates the **Products** worksheet from the Italian Restaurant Sales Analytics dataset. It verifies data quality, checks for missing values and duplicates, validates data types and product categories, and exports a cleaned dataset for further analysis in Python, SQL, and Power BI.

## Step 1 - Import Libraries

The first step is to import the Python libraries required for data manipulation and analysis.

```python
# Import pandas for data manipulation and analysis
import pandas as pd
```

## Step 2 - Load the Excel Workbook

Define the location of the Excel workbook.

```python
# Define the relative path to the Excel workbook
file_path = "../data/raw/restaurant-sales-analytics-data.xlsx"
```

## Step 3 - Read the Products Worksheet

Load the Products worksheet into a pandas DataFrame.

```python
# Load the Products worksheet into a pandas DataFrame
products = pd.read_excel(file_path, sheet_name="Products")
```

## Step 4 - Preview the Dataset

Display the first five rows of the Products dataset.

```python
# Display the first five rows
products.head()
```

## Step 5 - Inspect the Dataset Structure

Display the structure of the dataset, including data types, number of non-null values, and memory usage.

```python
# Display dataset information
products.info()
```

## Step 6 - Check Dataset Dimensions

Display the number of rows and columns in the dataset.

```python
# Display dataset dimensions
products.shape
```

## Step 7 - Check Missing Values

Identify whether the dataset contains any missing values in each column.

```python
# Check for missing values
products.isnull().sum()
```

## Step 8 - Check for Duplicate Rows

Identify whether the dataset contains duplicate records that may affect the analysis.

```python
# Check for duplicate rows
products.duplicated().sum()
```

## Step 9 - Display Column Names

Display all column names in the dataset.

```python
# Display column names
products.columns
```

## Step 10 - Display Summary Statistics

Display descriptive statistics for the numerical columns to better understand the distribution of the data.

```python
# Display summary statistics
products.describe()
```

## Step 11 - Check Unique Categories

Display all unique product categories to verify that they are correctly classified.

```python
# Display unique product categories
products["Category"].unique()
```

## Step 12 - Count Unique Categories

Count the number of unique product categories in the dataset.

```python
# Count unique product categories
products["Category"].nunique()
```

## Step 13 - Count Unique Products

Count the number of unique products in the dataset.

```python
# Count unique products
products["Product"].nunique()
```

## Step 14 - Display Data Types

Display the data types of all columns.

```python
# Display data types
products.dtypes
```

## Step 15 - Check for Negative Sales

Verify whether the dataset contains any negative sales values.

```python
# Check for negative sales
(products["Sales"] < 0).sum()
```

## Step 16 - Check for Zero Sales

Verify whether the dataset contains any products with zero sales.

```python
# Check for zero sales
(products["Sales"] == 0).sum()
```

## Step 17 - Check for Negative Quantity

Verify whether the dataset contains any negative quantities.

```python
# Check for negative quantities
(products["Quantity"] < 0).sum()
```

## Step 18 - Check for Zero Quantity

Verify whether the dataset contains any products with zero quantity sold.

```python
# Check for zero quantities
(products["Quantity"] == 0).sum()
```

## Step 19 - Create a Clean Copy

Create a copy of the cleaned dataset to preserve the original data before exporting.

```python
# Create a clean copy of the dataset
products_cleaned = products.copy()
```

## Step 20 - Verify the Clean Dataset

Display the first rows of the cleaned dataset to verify that it is ready for export.

```python
# Preview the cleaned dataset
products_cleaned.head()
```

## Step 21 - Export the Clean Dataset

Export the cleaned Products dataset to the cleaned data folder.

```python
# Export the cleaned dataset
products_cleaned.to_excel(
    "../data/cleaned/products_cleaned.xlsx",
    index=False
)

print("Products dataset exported successfully!")
```

## Step 22 - Cleaning Summary

Display a summary of the data cleaning process.

```python
print("========== CLEANING SUMMARY ==========")

print(f"Rows: {products.shape[0]}")
print(f"Columns: {products.shape[1]}")

print(f"Missing values: {products.isnull().sum().sum()}")
print(f"Duplicate rows: {products.duplicated().sum()}")

print(f"Unique Categories: {products['Category'].nunique()}")
print(f"Unique Products: {products['Product'].nunique()}")

print(f"Negative Sales: {(products['Sales'] < 0).sum()}")
print(f"Zero Sales: {(products['Sales'] == 0).sum()}")

print(f"Negative Quantity: {(products['Quantity'] < 0).sum()}")
print(f"Zero Quantity: {(products['Quantity'] == 0).sum()}")

print("Clean dataset exported successfully.")
print("======================================")
```
