# Italian Restaurant Sales Analytics

## 01 - Sales Data Cleaning and Preparation

### Objective

This notebook cleans and validates the **Sales** worksheet from the Italian Restaurant Sales Analytics dataset. It verifies data quality, checks for missing values and duplicates, validates data types, and exports a cleaned dataset for further analysis in Python, SQL, and Power BI.

## Step 1 - Import Libraries

The first step is to import the Python libraries required for data manipulation and analysis.

```python
# Import pandas for data manipulation and analysis
import pandas as pd
```

## Step 2 - Load the Excel Workbook

The next step is to define the location of the Excel file and load the workbook into Python.

```python
# Define the relative path to the Excel workbook
file_path = "../data/raw/restaurant-sales-analytics-data.xlsx"
```

## Step 3 - Read the Sales Worksheet

Load the Sales worksheet into a pandas DataFrame.

## Installing Required Dependencies (If Needed)

This project reads Excel files using the `openpyxl` engine.

```python
%pip install openpyxl
```

```python
# Load the Sales worksheet into a pandas DataFrame
sales = pd.read_excel(file_path, sheet_name="Sales")
```

## Step 4 - Preview the Dataset

Display the first five rows of the dataset.

```python
# Display the first five rows
sales.head()
```

## Step 5 - Inspect the Dataset Structure

Display the structure of the dataset, including data types, number of non-null values, and memory usage.

```python
# Display dataset information
sales.info()
```

## Step 6 - Check Dataset Dimensions

Display the number of rows and columns in the dataset.

```python
# Display dataset dimensions
sales.shape
```

## Step 7 - Check Missing Values

Identify whether the dataset contains any missing values in each column.

```python
# Check for missing values
sales.isnull().sum()
```

## Step 8 - Check for Duplicate Rows

Identify whether the dataset contains duplicate records that may affect the analysis.

```python
# Check for duplicate rows
sales.duplicated().sum()
```

## Step 9 - Display Column Names

Display all column names in the dataset.

```python
# Display column names
sales.columns
```

## Step 10 - Display Summary Statistics

Display descriptive statistics for the numerical columns to better understand the distribution of the data.

```python
# Display summary statistics
sales.describe()
```

## Step 11 - Check Unique Categories

Display the unique values in the Category column to verify that all expected categories are present.

```python
# Display unique categories
sales["Category"].unique()
```

## Step 12 - Count Unique Categories

Count the number of unique categories in the dataset.

```python
# Count unique categories
sales["Category"].nunique()
```

## Step 13 - Count Unique Dates

Count the number of unique dates in the dataset.

```python
# Count unique dates
sales["Date"].nunique()
```

## Step 14 - Count Unique Sales Values

Count the number of unique sales values in the dataset.

```python
# Count unique sales values
sales["Sales"].nunique()
```

## Step 15 - Display Data Types

Display the data type of each column individually.

```python
# Display data types
sales.dtypes
```

## Step 16 - Check for Negative Sales

Verify whether the dataset contains any negative sales values.

```python
# Check for negative sales values
(sales["Sales"] < 0).sum()
```

## Step 17 - Check for Zero Sales

Verify whether the dataset contains sales values equal to zero.

```python
# Check for zero sales values
(sales["Sales"] == 0).sum()
```

## Step 18 - Verify Date Range

Display the earliest and latest dates in the dataset.

```python
# Display minimum and maximum dates
sales["Date"].min(), sales["Date"].max()
```

## Step 19 - Sort the Dataset by Date

Sort the dataset in chronological order.

```python
# Sort the dataset by date
sales = sales.sort_values(by="Date")
```

## Step 20 - Reset the Index

Reset the DataFrame index after sorting.

```python
# Reset the DataFrame index
sales.reset_index(drop=True, inplace=True)
```

## Step 21 - Preview the Clean Dataset

Display the first rows of the cleaned dataset.

```python
# Preview cleaned dataset
sales.head()
```

## Step 22 - Export the Clean Dataset

Export the cleaned dataset as a new Excel file for future analysis.

```python
# Export cleaned Sales dataset
sales.to_excel(
    "../data/cleaned/sales_cleaned.xlsx",
    index=False
)
```

## Step 23 - Cleaning Summary

Display a summary of the data quality checks performed during the cleaning process.

```python
print("========== CLEANING SUMMARY ==========")

print(f"Rows: {sales.shape[0]}")
print(f"Columns: {sales.shape[1]}")
print(f"Missing values: {sales.isnull().sum().sum()}")
print(f"Duplicate rows: {sales.duplicated().sum()}")
print(f"Negative sales: {(sales['Sales'] < 0).sum()}")
print(f"Zero sales: {(sales['Sales'] == 0).sum()}")
print(f"Date range: {sales['Date'].min().date()} to {sales['Date'].max().date()}")

print("Clean dataset exported successfully.")
print("======================================")
```
