# Italian Restaurant Sales Analytics

## 03 - Payment Methods Data Cleaning and Preparation

### Objective

This notebook cleans and validates the **Payment Methods** worksheet from the Italian Restaurant Sales Analytics dataset. It verifies data quality, checks for missing values and duplicates, validates data types, and exports a cleaned dataset for further analysis in Python, SQL, and Power BI.

## Step 1 - Import Libraries

The first step is to import the Python libraries required for data manipulation and analysis.

```python
# Import pandas for data manipulation and analysis
import pandas as pd
```

## Step 2 - Load the Excel Workbook

The next step is to define the location of the Excel file and load the workbook into Python.

```python
# Define the relative path to the Excel workbook
file_path = "../data/raw/restaurant-sales-analytics-data.xlsx"
```

## Step 3 - Read the Payment Methods Worksheet

Load the Payment Methods worksheet into a pandas DataFrame.

```python
# Load the Payment Methods worksheet into a pandas DataFrame
payment_methods = pd.read_excel(file_path, sheet_name="Payment Methods")
```

## Step 4 - Preview the Dataset

Display the first five rows of the dataset.

```python
# Display the first five rows
payment_methods.head()
```

## Step 5 - Inspect the Dataset Structure

Display the structure of the dataset, including data types, number of non-null values, and memory usage.

```python
# Display dataset information
payment_methods.info()
```

## Step 6 - Check Dataset Dimensions

Display the number of rows and columns in the dataset.

```python
# Display dataset dimensions
payment_methods.shape
```

## Step 7 - Check Missing Values

Identify whether the dataset contains any missing values in each column.

```python
# Check for missing values
payment_methods.isnull().sum()
```

## Step 8 - Check for Duplicate Rows

Identify whether the dataset contains duplicate records that may affect the analysis.

```python
# Check for duplicate rows
payment_methods.duplicated().sum()
```

## Step 9 - Display Column Names

Display all column names in the dataset.

```python
# Display column names
payment_methods.columns
```

## Step 10 - Display Summary Statistics

Display descriptive statistics for the numerical columns to better understand the distribution of the data.

```python
# Display summary statistics
payment_methods.describe()
```

## Step 11 - Check Unique Payment Methods

Display the unique values in the Payment Methods column to verify that all expected categories are present.

```python
# Display unique Payment Methods
payment_methods["Payment Method"].unique()
```

## Step 12 - Count Unique Payment Methods

Count the number of unique Payment Methods in the dataset.

```python
# Count unique Payment Methods
payment_methods["Payment Method"].nunique()
```

## Step 13 - Count Unique Dates

Count the number of unique dates in the dataset.

```python
# Count unique dates
payment_methods["Date"].nunique()
```

## Step 14 - Count Unique Transactions

Count the number of unique transactions in the dataset.

```python
# Count unique transactions
payment_methods["Transactions"].nunique()
```

## Step 15 - Display Data Types

Display the data type of each column individually.

```python
# Display data types
payment_methods.dtypes
```

## Step 16 - Check for Negative Transaction Values

Verify whether the dataset contains any negative transaction values.

```python
# Check for negative transaction values
(payment_methods["Transactions"] < 0).sum()
```

## Step 17 - Check for Zero Transaction Values

Verify whether the dataset contains Transaction Values equal to zero.

```python
# Check for zero transaction values
(payment_methods["Transactions"] == 0).sum()
```

## Step 18 - Verify Date Range

Display the earliest and latest dates in the dataset.

```python
# Display minimum and maximum dates
payment_methods["Date"].min(), payment_methods["Date"].max()
```

## Step 19 - Sort the Dataset by Date

Sort the dataset in chronological order.

```python
# Sort the dataset by date
payment_methods = payment_methods.sort_values(by="Date")
```

## Step 20 - Reset the Index

Reset the DataFrame index after sorting.

```python
# Reset the DataFrame index
payment_methods.reset_index(drop=True, inplace=True)
```

## Step 21 - Preview the Clean Dataset

Display the first rows of the cleaned dataset.

```python
# Preview cleaned dataset
payment_methods.head()
```

## Step 22 - Export the Clean Dataset

Export the cleaned dataset as a new Excel file for future analysis.

```python
# Export cleaned Payment Methods dataset
payment_methods.to_excel(
    "../data/cleaned/payment_methods_cleaned.xlsx",
    index=False
)
```

## Step 23 - Cleaning Summary

Display a summary of the data quality checks performed during the cleaning process.

```python
print("========== CLEANING SUMMARY ==========")

print(f"Rows: {payment_methods.shape[0]}")
print(f"Columns: {payment_methods.shape[1]}")
print(f"Missing values: {payment_methods.isnull().sum().sum()}")
print(f"Duplicate rows: {payment_methods.duplicated().sum()}")
print(f"Unique Payment Methods: {payment_methods['Payment Method'].nunique()}")
print(f"Negative Transactions: {(payment_methods['Transactions'] < 0).sum()}")
print(f"Zero Transactions: {(payment_methods['Transactions'] == 0).sum()}")
print(f"Date range: {payment_methods['Date'].min().date()} to {payment_methods['Date'].max().date()}")

print("Clean dataset exported successfully.")
print("======================================")
```
