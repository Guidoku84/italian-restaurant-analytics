# Italian Restaurant Sales Analytics

## 04 - Employees Data Cleaning and Preparation

### Objective

This notebook cleans and validates the **Employees** worksheet from the Italian Restaurant Sales Analytics dataset. It verifies data quality, checks for missing values and duplicates, validates data types, and exports a cleaned dataset for further analysis in Python, SQL, and Power BI.

## Step 1 - Import Libraries

The first step is to import the Python libraries required for data manipulation and analysis.

```python
# Import pandas for data manipulation and analysis
import pandas as pd
```

## Step 2 - Load the Excel Workbook

The next step is to define the location of the Excel file and load the workbook into Python.

```python
# Define the relative path to the Excel workbook
file_path = "../data/raw/restaurant-sales-analytics-data.xlsx"
```

## Step 3 - Read the Employees Worksheet

Load the Employees worksheet into a pandas DataFrame.

```python
# Load the Employees worksheet into a pandas DataFrame
employees = pd.read_excel(file_path, sheet_name="Employees")
```

## Step 4 - Preview the Dataset

Display the first five rows of the dataset.

```python
# Display the first five rows
employees.head()
```

## Step 5 - Inspect the Dataset Structure

Display the structure of the dataset, including data types, number of non-null values, and memory usage.

```python
# Display dataset information
employees.info()
```

## Step 6 - Check Dataset Dimensions

Display the number of rows and columns in the dataset.

```python
# Display dataset dimensions
employees.shape
```

## Step 7 - Check Missing Values

Identify whether the dataset contains any missing values in each column.

```python
# Check for missing values
employees.isnull().sum()
```

## Step 8 - Check for Duplicate Rows

Identify whether the dataset contains duplicate records that may affect the analysis.

```python
# Check for duplicate rows
employees.duplicated().sum()
```

## Step 9 - Display Column Names

Display all column names in the dataset.

```python
# Display column names
employees.columns
```

## Step 10 - Display Summary Statistics

Display descriptive statistics for the numerical columns to better understand the distribution of the data.

```python
# Display summary statistics
employees.describe()
```

## Step 11 - Check Unique Employees

Display the unique values in the Employees column to verify that all expected categories are present.

```python
# Display unique Employees
employees["Employee"].unique()
```

## Step 12 - Count Unique Employees

Count the number of unique Employees in the dataset.

```python
# Count unique Employees
employees["Employee"].nunique()
```

## Step 13 - Count Unique Quantities

Count the number of unique quantities in the dataset.

```python
# Count unique quantity
employees["Quantity"].nunique()
```

## Step 14 - Count Unique Sales Values

Count the number of unique Sales Values in the dataset.

```python
# Count unique sales values
employees["Sales"].nunique()
```

## Step 15 - Display Data Types

Display the data type of each column individually.

```python
# Display data types
employees.dtypes
```

## Step 16 - Check for Negative Sales

Verify whether the dataset contains any negative sales values.

```python
# Check for negative sales values
(employees["Sales"] < 0).sum()
```

## Step 17 - Check for Zero Sales Values

Verify whether the dataset contains Sales Values equal to zero.

```python
# Check for zero sales values
(employees["Sales"] == 0).sum()
```

## Step 18 - Check for Negative Quantities

Verify whether the dataset contains negative quantities.

```python
(employees["Quantity"] < 0).sum()
```

## Step 19 - Check for Zero Quantities

Verify whether the dataset contains zero quantities.

```python
(employees["Quantity"] == 0).sum()
```

## Step 20 - Create a Clean Copy

```python
# Reset the DataFrame index
employees_cleaned = employees.copy()
```

## Step 21 - Preview the Clean Dataset

Display the first rows of the cleaned dataset.

```python
# Preview cleaned dataset
employees.head()
```

## Step 22 - Export the Clean Dataset

Export the cleaned dataset as a new Excel file for future analysis.

```python
# Export cleaned Employees dataset
employees.to_excel(
    "../data/cleaned/employees_cleaned.xlsx",
    index=False
)
```

## Step 23 - Cleaning Summary

Display a summary of the data quality checks performed during the cleaning process.

```python
print("========== CLEANING SUMMARY ==========")

print(f"Rows: {employees.shape[0]}")
print(f"Columns: {employees.shape[1]}")

print(f"Missing values: {employees.isnull().sum().sum()}")
print(f"Duplicate rows: {employees.duplicated().sum()}")

print(f"Unique Employees: {employees['Employee'].nunique()}")

print(f"Negative Sales: {(employees['Sales'] < 0).sum()}")
print(f"Zero Sales: {(employees['Sales'] == 0).sum()}")

print(f"Negative Quantity: {(employees['Quantity'] < 0).sum()}")
print(f"Zero Quantity: {(employees['Quantity'] == 0).sum()}")

print("Clean dataset exported successfully.")
print("======================================")
```
