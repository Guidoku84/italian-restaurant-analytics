# SQL Database Setup

This notebook creates the SQLite database used for the SQL analysis of the Italian Restaurant Analytics project.

The cleaned datasets are loaded into SQLite tables so they can later be analyzed using SQL queries.

```python
# Import the libraries required to work with data and SQLite databases

import pandas as pd
import sqlite3
```

## Create Database Connection

A connection is created to the SQLite database stored inside the `sql` folder of the project.

```python
# Create a connection to the SQLite database inside the sql folder

conn = sqlite3.connect("../sql/restaurant_analytics.db")

print("Database connection created successfully.")
```

## Load Cleaned Datasets

The cleaned Excel datasets are loaded into pandas DataFrames before being transferred into SQLite tables.

```python
# Load the cleaned datasets from the data/cleaned folder

sales_df = pd.read_excel("../data/cleaned/sales_cleaned.xlsx")
promotions_df = pd.read_excel("../data/cleaned/promotions_cleaned.xlsx")
payment_methods_df = pd.read_excel("../data/cleaned/payment_methods_cleaned.xlsx")
employees_df = pd.read_excel("../data/cleaned/employees_cleaned.xlsx")
products_df = pd.read_excel("../data/cleaned/products_cleaned.xlsx")

print("Cleaned datasets loaded successfully.")
```

## Verify Dataset Dimensions

The dimensions of each cleaned dataset are checked to confirm that the files were loaded correctly before creating the SQLite tables.

```python
# Display the number of rows and columns in each dataset

print("Sales:", sales_df.shape)
print("Promotions:", promotions_df.shape)
print("Payment Methods:", payment_methods_df.shape)
print("Employees:", employees_df.shape)
print("Products:", products_df.shape)
```

## Create SQLite Tables

The cleaned datasets are transferred into the SQLite database as separate tables for SQL analysis.

```python
# Transfer each cleaned DataFrame into a SQLite table

sales_df.to_sql("sales", conn, if_exists="replace", index=False)
promotions_df.to_sql("promotions", conn, if_exists="replace", index=False)
payment_methods_df.to_sql("payment_methods", conn, if_exists="replace", index=False)
employees_df.to_sql("employees", conn, if_exists="replace", index=False)
products_df.to_sql("products", conn, if_exists="replace", index=False)

print("SQLite tables created successfully.")
```

## Verify SQLite Tables

The database is checked to confirm that all cleaned datasets were successfully created as SQLite tables.

```python
# Display all tables stored in the SQLite database

tables = pd.read_sql_query(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)

tables
```

## Inspect Table Schemas

The structure of each SQLite table is inspected to verify column names and data types before creating the SQL schema script.

```python
# Display the column names and SQLite data types for each table

table_names = [
    "sales",
    "promotions",
    "payment_methods",
    "employees",
    "products"
]

for table in table_names:
    print(f"\n--- {table.upper()} ---")
    schema = pd.read_sql_query(f"PRAGMA table_info({table});", conn)
    print(schema[["name", "type"]].to_string(index=False))
```
